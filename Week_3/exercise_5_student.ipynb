## Task 1: Create a boxfilter with the size 5X5 and apply it on an image

Don't forget to normalize! use : filter_gaussian.png; 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('filter_gaussian.png', 0)

# ── 5x5 Box Filter ────────────────────────────────────────────────
# Each output pixel = average of its 5x5 neighborhood
# Dividing by 25 normalizes the kernel so brightness is preserved
kernel = np.ones((5, 5), np.float32) / 25.0
filtered_img = cv2.filter2D(img, -1, kernel)

# ── Plot ──────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('5×5 Box Filter', fontsize=13, fontweight='bold')

axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original')
axes[0].axis('off')

axes[1].imshow(filtered_img, cmap='gray')
axes[1].set_title('Box Filter (5×5)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

# Task 2: Create a gaussianfilter with several sizes 3x3,7x7, 15x15

What happens to an image by using this several sizes.use : filter_gaussian.png; You can use : cv2.GaussianBlur(img,kernelsize,0)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('filter_gaussian.png', 0)
blur_3 = cv2.GaussianBlur(img, (3, 3), 0)
blur_7 = cv2.GaussianBlur(img, (7, 7), 0)
blur_15 = cv2.GaussianBlur(img, (15, 15), 0)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.title('Gaussian Blur (3x3)')
plt.imshow(blur_3, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.title('Gaussian Blur (7x7)')
plt.imshow(blur_7, cmap='gray')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.title('Gaussian Blur (15x15)')
plt.imshow(blur_15, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Task 2 — Gaussian Filter

As the kernel size increases, more neighboring pixels are included in the
weighted average — producing a progressively stronger blur:

- **3×3:** subtle smoothing, fine details mostly preserved
- **7×7:** moderate blur, small structures start to disappear
- **15×15:** heavy blur, only large structures remain visible

Unlike the box filter, Gaussian weighting gives more influence to nearby
pixels and less to distant ones — producing a smoother, more natural-looking
result that better matches how optics and biological vision work.

## Task 3: Using a Median filter on a salt and pepper image

Use  cv2.medianBlur(img,kernelsize); Use these images: noisy_multiply; noisy_add;

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ── Load images ───────────────────────────────────────────────────
img_add  = cv2.imread('saltandpepper_add.png', 0)
img_mult = cv2.imread('saltandpepper_multiply.png', 0)

# Check what we loaded
print("Add image - min:", img_add.min(), "max:", img_add.max(), "shape:", img_add.shape)
print("Mult image - min:", img_mult.min(), "max:", img_mult.max(), "shape:", img_mult.shape)

# ── Apply Median Filter ───────────────────────────────────────────
filtered_add  = cv2.medianBlur(img_add, 5)
filtered_mult = cv2.medianBlur(img_mult, 5)

# ── Plot ──────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Median Filter on Salt & Pepper Noise', fontsize=13, fontweight='bold')

axes[0, 0].imshow(img_add, cmap='gray', vmin=0, vmax=255)
axes[0, 0].set_title('Salt & Pepper Add (Original)')
axes[0, 0].axis('off')

axes[0, 1].imshow(filtered_add, cmap='gray', vmin=0, vmax=255)
axes[0, 1].set_title('After Median Filter (Add)')
axes[0, 1].axis('off')

axes[1, 0].imshow(img_mult, cmap='gray', vmin=0, vmax=255)
axes[1, 0].set_title('Salt & Pepper Multiply (Original)')
axes[1, 0].axis('off')

axes[1, 1].imshow(filtered_mult, cmap='gray', vmin=0, vmax=255)
axes[1, 1].set_title('After Median Filter (Multiply)')
axes[1, 1].axis('off')

plt.tight_layout()
plt.show()
```
